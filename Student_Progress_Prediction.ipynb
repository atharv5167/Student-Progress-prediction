```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('student_exam_data_new.csv')
```

```python
import matplotlib as plt
```

```python
import matplotlib.pyplot as plt

scatter = plt.scatter(df['Study Hours'], df['Previous Exam Score'], c=df['Pass/Fail'], cmap='coolwarm')

# Add color scale box
plt.colorbar(scatter, label="Pass/Fail (0 = Fail, 1 = Pass)")

# Labels
plt.xlabel("Study Hours")
plt.ylabel("Previous Exam Score")
plt.title("Study Hours vs Previous Exam Score (Pass/Fail)")

plt.show()
```

```python
X = df.iloc[:,0:2]
Y = df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.1)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
```

```python
X_test = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
```

```python
clf = LogisticRegression()
```

```python
clf.fit(X_train,Y_train)
```

```python
Y_pred = clf.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(Y_test,Y_pred)
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(X_train, Y_train.values, clf=clf, legend=2)
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, Y_train)
Y_pred = knn.predict(X_test)
accuracy_score(Y_test, Y_pred)
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(Y_test, Y_pred)  
```
